### Exercise 1.1: Counting Words

Given the sentence from the slides: *"They picnicked by the pool, then lay back on the grass and looked at the stars."*

Write Python code to:
1. Count words excluding punctuation
2. Count words including punctuation as separate tokens
3. Count unique word types (case-insensitive)

````python
# YOUR CODE HERE
import re

sentence = "They picnicked by the pool, then lay back on the grass and looked at the stars."

# 1. Count words excluding punctuation
clean_sentence = re.findall(r'\w+', sentence) # Extracting the punctuation with re
print(f"Total words: {len(clean_sentence)}")

# 2. Count words including punctuation
total_sentence = re.findall(r'(\w+|[^\w\s])', sentence) # Extracting words and punctuation separately
print(f"Total words counting punctuation: {len(total_sentence)}") # Punctuation counts as separate words

# 3. Count unique types (case-insensitive)
unique_types = len(set(word.lower() for word in clean_sentence)) # Converting to lower case to avoid duplicates due to case
print(f"Unique types: {unique_types}")
````

### Exercise 1.2: Handling Disfluencies

The slides show the utterance: *"I do uh main- mainly business data processing"*

Write code to:
1. Count all tokens including disfluencies
2. Remove filled pauses (uh, um) and fragments (words ending with -)
3. Count "clean" words

````python
# YOUR CODE HERE
utterance = "I do uh main- mainly business data processing"

# 1. Count all tokens including disfluencies
tokens = utterance.split()
print(f"Total tokens: {len(tokens)}")

# 2. Remove filled pauses (uh, um) and fragments (words ending with -)
clean_tokens = re.sub(r'(uh|um|\w+-\s)', '', utterance) # Removing filled pauses and fragments and substituting with empty string
clean_tokens = clean_tokens.split() # Tokenizing the cleaned utterance for clearing residual spaces
print(" ".join(clean_tokens)) # Print cleaned utterance with spaces

#3. Count "clean" words
print(f"Number of clean words: {len(clean_tokens)}") # Counting the clean words
````

### Exercise 2.1: Basic Pattern Matching

Using the regex patterns from the slides, write code to:
1. Find all words starting with capital letters
2. Find all digits in a text
3. Find words that are NOT capitalized

````python
# YOUR CODE HERE
text = "Chapter 1: Down the Rabbit Hole. Alice was 7 years old in 1865."

capital_words = re.findall(r'\b[A-Z]\w+', text) # Only words that start with capital letter
print(capital_words)

digits = re.findall(r'\d+', text) # All digit sequences
print(digits)

not_capital_words = re.findall(r'\b[a-z]\w+', text) # Only words that start with lowercase letter
print(not_capital_words)
````

### Exercise 2.2: Kleene Star and Plus

The slides explain Kleene operators (* and +).

Write patterns to match:
1. "baa" followed by zero or more 'a's (baa, baaa, baaaa...)
2. One or more digits
3. Any sequence of characters (wildcard)

````python
# YOUR CODE HERE
test_strings = ["ba", "baa", "baaa", "baaaa", "baaaaa"]

kleene = []
digits = []
characters = []

for string in test_strings:
    if re.search(r'baaa*', string):
        kleene.append(string)
    if re.search(r'\d+', string):
        digits.append(string)
    if re.search(r'.*', string):
        characters.append(string)

print(kleene)
print(digits)
print(characters)
````

### Exercise 2.3: Anchors and Word Boundaries

Write patterns using anchors (^, $, \b) to:
1. Match lines starting with a capital letter
2. Match lines ending with a period
3. Find the word "the" as a complete word (not in "other" or "there")

````python
# YOUR CODE HERE
lines = [
    "The quick brown fox",
    "jumped over the lazy dog.",
    "other animals watched.",
    "There was nothing else."
]

capital_lines = []
period_lines = []

for line in lines:
    if re.search(r'^[A-Z]', line):
        capital_lines.append(line)
    if re.search(r'\.$', line):
        period_lines.append(line)
    
    found_the = re.findall(r'\b[Tt]he\b', line)
    if found_the:
        print(f"The found in this line: {found_the}")

print(capital_lines)
print(period_lines)
````

### Exercise 2.5: ELIZA-style Pattern Matching

The slides show ELIZA, an early chatbot using regex.
Implement simple ELIZA rules for:
1. "I'm [emotion]" -> "WHY DO YOU THINK YOU ARE [emotion]"
2. "I need [something]" -> "WHAT WOULD IT MEAN IF YOU GOT [something]"
3. Any input with "always" -> "CAN YOU THINK OF A SPECIFIC EXAMPLE?"

````python
# YOUR CODE HERE
def simple_eliza(user_input):
    response = []

    for line in user_input:
        if re.search(r"I'm \S+", line):
            response.append(re.sub(r"I'm (\S+)", r"WHY DO YOU THINK YOU ARE \1", line).upper()) # Converting the responses into uppercase

        elif re.search(r"I need .*", line): # Searching for I need followed by anything
            response.append(re.sub(r"I need (\S+)", r"WHAT WOULD IT MEAN IF YOU GOT \1", line).upper())

        elif re.search(r"always", line): # Searching for always
            response.append("CAN YOU THINK OF A SPECIFIC EXAMPLE?")

    return response

# Test
test_inputs = [
    "I'm depressed",
    "I need a vacation", 
    "They always ignore me"
]

responses = simple_eliza(test_inputs)
print(responses)
````
