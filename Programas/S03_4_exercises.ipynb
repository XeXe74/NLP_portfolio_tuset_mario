### Exercise 1.1: Types vs Instances

The slides distinguish between types and instances.
For the given text, calculate:
1. Number of instances (total tokens)
2. Number of types (vocabulary size |V|)
3. Type-token ratio

```python
# YOUR CODE HERE
text = "the cat sat on the mat the cat was fat"

tokens = text.split() # Tokenizate the text
instances = len(tokens) # Number of total words

num_types = len(set(tokens)) # Number of unique words

type_token_ratio = instances / num_types # Type-token ratio

print(f"Number of instances: {instances}")
print(f"Number of types: {num_types}")
print(f"Type-token ratio: {type_token_ratio}")
```

### Exercise 1.2: Heaps' Law Demonstration

The slides mention Heaps' Law: vocabulary size grows with √N.

Generate text of increasing length and observe vocabulary growth.

```python
# YOUR CODE HERE
import random

# Use a simple word list to simulate text
word_list = ['the', 'a', 'is', 'are', 'was', 'were', 'be', 'been',
             'cat', 'dog', 'bird', 'fish', 'tree', 'house', 'car',
             'run', 'walk', 'jump', 'eat', 'sleep', 'read', 'write',
             'big', 'small', 'fast', 'slow', 'red', 'blue', 'green']


text_lengths = [5, 10, 50, 100, 200, 500] # Text lengths for the generated text

for length in text_lengths:
    generated_words = []
    
    for _ in range(length):
        generated_words.append(random.choice(word_list)) # A random word from the word list is taken for the generated text

    vocabulary_size = len(set(generated_words))

    print(f"Vocabulary size for {length} text length: {vocabulary_size}")
```

### Exercise 1.3: BPE Simulation

The slides explain Byte Pair Encoding (BPE).
Implement a simple BPE token learner that:
1. Starts with character vocabulary
2. Finds most frequent adjacent pair
3. Merges them into a new token
4. Repeats k times

```python
import re
from collections import Counter

def simple_bpe_word_level(corpus_str, num_merges):
    vocab = [list(word) + ['</w>'] for word in corpus_str.split()] # Adding end-of-word token
    
    print(f"Initial vocabulary: {vocab}")

    for i in range(num_merges):
        pairs = Counter() # Adding a counter for pairs of symbols
        
        for word in vocab:
            for j in range(len(word) - 1):
                pairs[(word[j], word[j + 1])] += 1 # Count pairs of symbols
        
        if not pairs:
            break
            
        best_pair = pairs.most_common(1)[0][0] # Get the most common pair
        print(f"Merge {i+1}: {best_pair}")
        
        # Applying the merge to the vocabulary
        new_vocab = []
        for word in vocab:
            new_word = []
            j = 0
            while j < len(word):
                if j < len(word) - 1 and (word[j], word[j + 1]) == best_pair: # If the best pair is found
                    new_word.append(''.join(best_pair)) # Merge the pair
                    j += 2
                else:
                    new_word.append(word[j]) # Keep the symbol as is
                    j += 1
            new_vocab.append(new_word)
        vocab = new_vocab
        
    return vocab

# Prueba con tu string
corpus = "low lower newest widest"
final_vocab = simple_bpe_word_level(corpus, 5)
```

### Exercise 2.1: Lookahead Assertions

The slides introduce lookahead: (?=pattern) and (?!pattern)

Write patterns to:
1. Find words followed by a comma (without capturing comma)
2. Find first word of line only if it doesn't start with 'T'

```python
# YOUR CODE HERE
text = "The quick, brown fox jumps over the lazy dog."

commas = re.findall(r'\w+(?=,)', text) # Lookahead for searching for a comma
first_word = re.findall(r'^[^Tt]\w*', text) # ^ For the line start

print(commas)
print(first_word)
```
