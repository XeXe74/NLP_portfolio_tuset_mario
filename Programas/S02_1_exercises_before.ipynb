```python
# Exercise 1: Make Plurals
# Write a program to generate plural forms of English nouns from a given list of singular nouns.
# Simplification rules:
# 1. Add 's' to most words (e.g., "cat" -> "cats").
# 2. If the word ends with 'y' and is preceded by a consonant, replace 'y' with 'ies' (e.g., "baby" -> "babies").
# 3. If the word ends with 'f' or 'fe', replace it with 'ves' (e.g., "wolf" -> "wolves").
# 4. Add 'es' to words ending in 's', 'x', 'z', 'ch', or 'sh' (e.g., "box" -> "boxes").
# 5. For irregular nouns, handle exceptions (e.g., "child" -> "children", "man" -> "men").

nouns = ["cat", "baby", "wolf", "box", "child", "man", "mouse", "knife", "bush"]

def make_plurals(nouns):
    plural = []
    for noun in nouns:
        if noun[-1] == "y" and noun[-2] not in "aeiou":
            plural.append(noun[:-1] + "ies")
            
        elif noun[-1] == "f":
            plural.append(noun[:-1] + "ves")
            
        elif noun[-2:] == "fe":
            plural.append(noun[:-2] + "ves")
            
        elif noun[-1] in "sxz" or noun[-2:] in ["ch", "sh"]:
            plural.append(noun + "es")
            
        elif noun == "child":
            plural.append("children")
            
        elif noun == "man":
            plural.append("men")
            
        elif noun == "mouse":
            plural.append("mice")
        
        else:
            plural.append(noun + "s")
            
    return plural
            
        
```

```python
import re

def make_plurals_re(nouns):
    plural = []

    for noun in nouns:
        if noun == "child":
            plural.append("children")
        
        elif noun == "man":
            plural.append("men")

        elif noun == "mouse":
            plural.append("mice")

        elif re.search(r'[^aeiou]y$', noun):
            plural.append(re.sub(r'y$', 'ies', noun))

        elif re.search(r'f$', noun):
            plural.append(re.sub(r'f$', 'ves', noun))

        elif re.search(r'fe$', noun):
            plural.append(re.sub(r'fe$', 'ves', noun))

        elif re.search(r'(s|x|z|ch|sh)$', noun):
            plural.append(noun + "es")

        else:
            plural.append(noun + "s")

    return plural
```

```python
# plural = make_plurals(nouns)
plural = make_plurals_re(nouns)

print(plural)
```

```python
# Exercise 2: Extract Email Addresses
# Write a Python program to extract all email addresses from the following string.
text = "Please contact us at support@example.com, sales@example.org, or admin123@example.net for assistance."

def extract_email(text):
    emails = []
    for word in text.split():
        if "@" in word:
            emails.append(word.strip(",."))
            
    return emails
```

```python
def extract_email_re(text):
    
    return re.findall(r'\S+@\S+', text)
```

```python
# emails = extract_email(text)
emails = extract_email_re(text)

print(emails)
```

```python
# Exercise 3: Extract Hashtags
# From a tweet, extract all valid hashtags.
tweet = "Loving the #sunset at the #beach! But #123abc and ##double are #invalid#tags."

def extract_hashtag(text):
    hashtags = []
    counter = 0
    valid = True
    
    for word in text.split():
        for char in word:
            if char == "#":
                counter += 1
                
        if len(word) > 1 and word[1].isnumeric():
            valid = False
            
        if counter == 1 and valid and word[0] == "#":
            hashtags.append(word.strip(".,!"))
            
        counter = 0
        valid = True
            
    return hashtags          
            
```

```python
hashtags = extract_hashtag(tweet)

print(hashtags)
```
