### Exercise 1 (Easy)
Generate all bigrams and trigrams from a sentence.

```python
sentence = "I love natural language processing"

def get_ngrams(tokens, n):
    # Your implementation
    n_grams = []

    for i in range(len(tokens) - n + 1) : # We use -n + 1 to avoid index out of range
        n_grams.append(tuple(tokens[i:i+n])) # We use tuple to make it hashable and easier to work with
    
    return n_grams

tokens = sentence.lower().split()
# Get bigrams and trigrams
bigrams = get_ngrams(tokens, 2)
trigrams = get_ngrams(tokens, 3)
print("Bigrams:", bigrams)
print("Trigrams:", trigrams)
```

### Exercise 2 (Easy)
Calculate the probability of a bigram using Maximum Likelihood Estimation (MLE).

```python
corpus = ["i love nlp", "i love python", "i hate bugs", "nlp is great"]

# Calculate P(love | i) using MLE
# P(love | i) = count(i, love) / count(i)
tokens_corpus = [sentence.lower().split() for sentence in corpus] # Tokenize the corpus
count_i_love = 0
count_i = 0

for tokens in tokens_corpus:
    for i in range(len(tokens) - 1):
        if tokens[i] == "i": # Check if the current token is "i"
            count_i += 1
            if tokens[i+1] == "love": # Check if the next token is "love"
                count_i_love += 1

p_love_i = count_i_love / count_i # Calculate the probability
print(f"P(love | i) = {p_love_i:.3f}")
    
```
