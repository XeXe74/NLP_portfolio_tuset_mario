### Exercise 1 (Easy)
Create a simple RNN cell and pass a sequence through it.

```python
import torch
import torch.nn as nn

# Create RNN: input_size=10, hidden_size=20
# Pass a sequence of shape (seq_len=5, batch=1, input_size=10)
rnn = nn.RNN(input_size=10, hidden_size=20, batch_first=False)

x  = torch.randn(5, 1, 10)
h0 = torch.zeros(1, 1, 20)

rnn_output, rnn_hn = rnn(x, h0) # Get output and final hidden state

print(f"Input shape: {x.shape}")
print(f"Output shape: {rnn_output.shape}")
print(f"Final hidden (hn): {rnn_hn.shape}")
```

### Exercise 2 (Easy)
Create an LSTM and compare output shapes with vanilla RNN.

```python
# Create LSTM with same dimensions
# Note: LSTM returns (output, (h_n, c_n))
lstm = nn.LSTM(input_size=10, hidden_size=20, batch_first=False)

c0 = torch.zeros(1, 1, 20) # LSTM needs both h0 and c0

lstm_output, (lstm_hn, lstm_cn) = lstm(x, (h0, c0)) # Get output, final hidden state, and cell state

# Compare results
print("RNN:")
print(f"Output: {rnn_output.shape}")
print(f"HN: {rnn_hn.shape}")

print("\nLSTM:")
print(f"Output: {lstm_output.shape}")
print(f"HN: {lstm_hn.shape}")
print(f"CN: {lstm_cn.shape}")
```
