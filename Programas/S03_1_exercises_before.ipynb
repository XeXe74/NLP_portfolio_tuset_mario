# Text Tokenization Exercise

This exercise explores the challenges of splitting text into sentences and words when dealing with complex real-world text containing dates, amounts, URLs, emails, acronyms, and multi-word expressions.

## The Challenge

Given a text variable, split it into:
1. **Sentences** - logical units of meaning ending with terminal punctuation
2. **Words (tokens)** - individual meaningful units

```python
import re
import nltk
from nltk.tokenize import word_tokenize, RegexpTokenizer
from collections import Counter
from nltk.corpus import stopwords
import spacy
import string
```

```python
def split_sentences(text): # Done before class explanation
    words = []
    sentences = []
    
    words = re.findall(r'\S+', text) # Not complete
    print(words)
    
    print("\n")
    
    sentences = re.findall(r'', text) # Not done
    print(sentences)
    
nltk.download('punkt_tab') # Download punkt tokenizer for nltk
    
def split_sentences_nltk(text):
    """
    Function to split text into words and sentences using nltk.
    
    """
    
    sentences = nltk.sent_tokenize(text) # Sentence tokenization using nltk
    
    tknzr = RegexpTokenizer(pattern=r'[\w$.@%€:/-]+') # Word tokenization using regex
    tokens = tknzr.tokenize(text)

    # Print the results
    print(f"Total words: {len(tokens)}")
    print(tokens)
    
    print(f"\nTotal sentences: {len(sentences)}")
    for sent in sentences:
        print(sent)
        
nlp = spacy.load("en_core_web_sm") # Load spacy model

def split_sentences_spacy(text):
    """
    Function to split text into words and sentences using spacy.
    
    """
    
    doc = nlp(text) # Process the text with spacy
    
    tokens = [ # Tokenization with special cases
        token.text for token in doc 
        if not token.is_punct or token.text in ['$', '€', '%']
    ]
    
    clean_words = []
    i = 0
    
    while i < len(tokens): # Handle special cases
        current = tokens[i]
        
        if current in ['$', '€'] and i+1 < len(tokens) and tokens[i+1][0].isdigit(): # Money symbol before number
            clean_words.append(current + tokens[i+1])
            i += 2
            
        elif current[0].isdigit() and i+1 < len(tokens) and tokens[i+1] == '%': # Percentage symbol after number
            clean_words.append(current + tokens[i+1])
            i += 2
            
        else:
            clean_words.append(current)
            i += 1
    
    sentences = [sent.text for sent in doc.sents] # Sentence tokenization with spacy
    
    # Print the results
    print(f"Total words: {len(clean_words)}")
    print(clean_words)
    
    print(f"\nTotal sentences: {len(sentences)}")
    for sent in sentences:
        print(sent)
```

```python
# Sample text with challenging elements
text = """Dr. John Smith, Ph.D., earned $1,250.50 on Jan. 15, 2024, for his work at A.I. Corp. You can reach him at j.smith@ai-corp.co.uk or visit https://www.ai-corp.co.uk/team/dr-smith for more info. The U.S.A.-based company reported a 23.5% increase in Q3 revenue, totaling €2.5M."""

print("Original text:")
print(text)
print("\n")
split_sentences(text)
```

```python
split_sentences_nltk(text)
```

```python
split_sentences_spacy(text)
```

# Corpus Tokenization Exercise

This exercise explores the challenges of splitting words in large corpuses and find the most common words. 

## The Challenge

Given a file `shakes.txt` in the book folder. Find the words that are more common in Shakespeare's book. 

```python
nltk.download("stopwords") # Download stopwords for nltk

def find_common_words_nltk(filename):
    """
    Find most common words using NLTK tokenization and stop words.
    """
    with open(filename, 'r', encoding='utf-8') as file: # Open the file
        text = file.read().lower()
    
    tokens = word_tokenize(text) # Tokenize the text
    
    stop_words = set(stopwords.words('english')) # Get English stop words
    words = [word for word in tokens 
             if word.isalpha() and word not in stop_words]
    
    word_counts = Counter(words) # Count word frequencies
    
    # Print results
    print(f"Total words (without stop words): {len(words)}") # Total words count
    print(f"Unique words: {len(word_counts)}") # Unique words count
    print(f"\nTop 15 most common words:\n")
    top_15 = word_counts.most_common(15) # Get top 15 common words
    print(top_15)   
```

```python
find_common_words_nltk("books/shakes.txt")
```
