### Exercise 1 (Easy)
Load pre-trained Word2Vec embeddings and find similar words.

```python
import gensim.downloader as api

# Load pre-trained word2vec (google-news-300 or glove-wiki-gigaword-100)
# Find top 5 most similar words to 'king'

# Load pre-trained Word2Vec model (Google News, 300 dimensions)
model = api.load("word2vec-google-news-300")

# Find top 5 most similar words to 'king'
similar_words = model.most_similar("king", topn=5)

print("Top 5 words most similar to 'king':")
for word, score in similar_words:
    print(f"  {word}: {score:.4f}")

```

### Exercise 2 (Easy)
Perform word analogy: king - man + woman = ?

```python
# Use the model to solve: king - man + woman = ?
# Also try: paris - france + spain = ?

result1 = model.most_similar(positive=["king", "woman"], negative=["man"], topn=3) # Find top 3 results for the analogy
print("king - man + woman:")
for word, score in result1:
    print(f"  {word}: {score:.4f}")

result2 = model.most_similar(positive=["paris", "spain"], negative=["france"], topn=3)
print("\nparis - france + spain:")
for word, score in result2:
    print(f"  {word}: {score:.4f}")
```

### Exercise 3 (Medium)
Train your own Word2Vec model on a custom corpus.

```python
from gensim.models import Word2Vec

corpus = [
    ["the", "cat", "sat", "on", "the", "mat"],
    ["the", "dog", "ran", "in", "the", "park"],
    ["cats", "and", "dogs", "are", "pets"],
    ["the", "cat", "chased", "the", "dog"],
    ["pets", "need", "food", "and", "water"]
]

# Train Word2Vec model (vector_size=50, window=3, min_count=1)

# Train Word2Vec model
model_custom = Word2Vec(sentences=corpus, vector_size=50, window=3, min_count=1, workers=4)

# Most similar to 'cat'
print("Most similar to 'cat':")
for word, score in model_custom.wv.most_similar("cat", topn=3):
    print(f"  {word}: {score:.4f}")

# Vector for 'cat'
print("\nVector for 'cat':")
print(model_custom.wv["cat"])
```

### Exercise 4 (Medium)
Build a simple neural network for text classification using embeddings.

```python
import torch
import torch.nn as nn

class TextClassifier(nn.Module):
    """
    Simple text classification model using embeddings and mean pooling.
    """
    def __init__(self, vocab_size, embed_dim, num_classes):
        # Define: embedding layer, linear layer
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)
        self.fc = nn.Linear(embed_dim, num_classes)

    
    def forward(self, x):
        # Embed -> mean pooling -> classify
        embedded = self.embedding(x)
        pooled   = embedded.mean(dim=1)
        logits   = self.fc(pooled)
        return logits

# Test with dummy data
x = torch.randint(1, 1000, (8, 20))
labels = torch.randint(0, 4, (8,))

model   = TextClassifier(1000, 64, 4)
logits  = model(x) # Forward pass to get logits
preds   = logits.argmax(dim=1) # Get predicted class indices

print(f"Input shape  : {x.shape}")
print(f"Logits shape : {logits.shape}")
print(f"Predictions  : {preds.tolist()}")
print(f"Labels       : {labels.tolist()}")

# Training loop
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
criterion = nn.CrossEntropyLoss()

for epoch in range(5):
    optimizer.zero_grad()
    loss = criterion(model(x), labels)
    loss.backward()
    optimizer.step()
    print(f"Epoch {epoch+1}  loss: {loss.item():.4f}")
```
