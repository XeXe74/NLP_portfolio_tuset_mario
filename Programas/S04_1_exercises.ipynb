```python
from collections import defaultdict

phrases = [
    "I am Sam",
    "Sam I am",
    "I do not like green eggs and ham"
]

tokenized_corpus = [] # List to hold tokenized sentences

for phrase in phrases:
    # Split by whitespace and add tags
    tokens = ["<s>"] + phrase.split() + ["</s>"]
    tokenized_corpus.append(tokens)

# Display the processed data
print("Processed Corpus:")
for sent in tokenized_corpus:
    print(sent)
```

```python
unigram_counts = defaultdict(int) # Counts of individual words
bigram_counts = defaultdict(lambda: defaultdict(int)) # Counts of word pairs

# Iterate through each sentence in the corpus
for sentence in tokenized_corpus:
    for i in range(len(sentence) - 1): # Loop through words, stopping before the last word
        current_word = sentence[i]  # Current word // c(wi)
        next_word = sentence[i + 1] # Next word // c(wi+1)
        
        # Update counts
        unigram_counts[current_word] += 1
        bigram_counts[current_word][next_word] += 1

print("Training complete. Counts calculated.")
print(f"Unique words in context (unigrams): {len(unigram_counts)}")
```

```python
# Display the probabilities in a readable format
print(f"{'Bigram Model Probabilities':^40}")
print("=" * 50)
print(f"{'Bigram':<20} | {'Math':<10} | {'Result':<10}")
print("-" * 50)

# Iterate through the counts to calculate probabilities
for prev_word in bigram_counts:
    for next_word in bigram_counts[prev_word]:
        
        # Get the counts
        numerator = bigram_counts[prev_word][next_word]   # c(w_{i-1}, w_i)
        denominator = unigram_counts[prev_word]           # c(w_{i-1})
        
        # Calculate probability
        prob = numerator / denominator
        
        # Formatting for display
        label = f"P({next_word} | {prev_word})"
        fraction = f"{numerator}/{denominator}"
        
        print(f"{label:<20} | {fraction:<10} | {prob:.2f}")
```

```python
def get_sentence_probability(sentence):
    tokens = ["<s>"] + sentence.split() + ["</s>"]
    total_prob = 1.0
    print(f"\nCalculating probability for: '{sentence}'")
    
    for i in range(len(tokens) - 1):
        prev = tokens[i]
        curr = tokens[i+1]
        
        # Fetch probability (return 0 if the bigram was never seen)
        if prev in bigram_counts and curr in bigram_counts[prev]:
            count_bigram = bigram_counts[prev][curr]
            count_uni = unigram_counts[prev]
            p = count_bigram / count_uni
            print(f"  P({curr} | {prev}) = {p:.2f}")
        else:
            p = 0.0
            print(f"  P({curr} | {prev}) = 0.00 (Unseen bigram)")
            
        total_prob *= p
        
    return total_prob

# Test with a specific phrase
prob = get_sentence_probability("I am Sam")
print(f"Total Probability: {prob:.4f}")
```
